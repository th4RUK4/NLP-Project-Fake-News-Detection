# 06 — Model Evaluation

Aggregate saved metrics from the ML and DL notebooks. This notebook does not retrain models.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path.cwd(); METRICS = ROOT / "reports" / "metrics"
paths = [METRICS / "logistic_regression_metrics.csv", METRICS / "lstm_metrics.csv"]
frames=[]
for path in paths:
    if path.exists():
        frames.append(pd.read_csv(path))
    else:
        print("Not found yet:", path)
if not frames:
    raise FileNotFoundError("Run notebooks 04 and 05 first.")
comparison = pd.concat(frames, ignore_index=True)
comparison = comparison.drop_duplicates(subset=["model"], keep="last")
comparison.to_csv(METRICS / "model_comparison.csv", index=False)
comparison
```

```python
plot_data = comparison.set_index("model")[["accuracy", "precision", "recall", "f1_score", "roc_auc"]]
ax = plot_data.plot(kind="bar", figsize=(10, 6))
ax.set_ylim(0, 1.05); ax.set_ylabel("Score"); ax.set_title("Fake News Detection Model Comparison"); ax.legend(loc="lower right"); plt.tight_layout(); plt.savefig(ROOT / "reports" / "figures" / "model_comparison.png", dpi=150); plt.show()
```
