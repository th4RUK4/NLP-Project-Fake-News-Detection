# 02 — Data Preprocessing

Create the deterministic `clean_text` field from raw `content`. The preprocessing code is shared with inference through `src/preprocessing.py`.

```python
from pathlib import Path
import pandas as pd
import nltk

nltk.download("stopwords", quiet=True)
nltk.download("wordnet", quiet=True)
nltk.download("omw-1.4", quiet=True)

from src.preprocessing import preprocess_text

PROJECT_ROOT = Path.cwd()
RAW_PATH = PROJECT_ROOT / "data" / "raw" / "combined_news.csv"
EXISTING_PATH = PROJECT_ROOT / "data" / "processed" / "cleaned_news.csv"
OUTPUT_PATH = PROJECT_ROOT / "data" / "processed" / "cleaned_news.csv"

# Prefer the raw combined dataset when available; otherwise validate the existing cleaned dataset.
DATA_PATH = RAW_PATH if RAW_PATH.exists() else EXISTING_PATH
if not DATA_PATH.exists():
    raise FileNotFoundError("Put the source dataset in data/raw/combined_news.csv or provide data/processed/cleaned_news.csv.")

df = pd.read_csv(DATA_PATH)
if "content" not in df.columns or "label" not in df.columns:
    raise ValueError("Dataset must contain content and label columns.")

df["content"] = df["content"].fillna("").astype(str)
df["clean_text"] = df["content"].map(preprocess_text)
df["label"] = df["label"].astype(int)
OUTPUT_PATH.parent.mkdir(parents=True, exist_ok=True)
df[["content", "clean_text", "label"]].to_csv(OUTPUT_PATH, index=False)
print("Saved:", OUTPUT_PATH)
print("Shape:", df.shape)
```
