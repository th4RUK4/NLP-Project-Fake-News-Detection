# 03 — Feature Engineering & Reproducible Split

Create one stratified train/test split used by both Logistic Regression and LSTM. No vocabulary or TF-IDF statistics are learned here.

```python
from pathlib import Path
import pandas as pd
from sklearn.model_selection import train_test_split

ROOT = Path.cwd()
DATA = ROOT / "data" / "processed" / "cleaned_news.csv"
SPLIT = ROOT / "data" / "processed" / "model_split.csv"

df = pd.read_csv(DATA)
if not {"clean_text", "label"}.issubset(df.columns):
    raise ValueError("cleaned_news.csv must contain clean_text and label")

indices = df.index.to_numpy()
train_idx, test_idx = train_test_split(indices, test_size=0.20, random_state=42, stratify=df["label"])
split = pd.DataFrame({"row_index": indices, "split": "train"})
split.loc[split["row_index"].isin(test_idx), "split"] = "test"
split.to_csv(SPLIT, index=False)

print(split["split"].value_counts())
print("Saved:", SPLIT)
```

```python
train = df.loc[train_idx]
test = df.loc[test_idx]
print("Training:", len(train))
print("Testing:", len(test))
print("Train labels:\n", train["label"].value_counts(normalize=True).sort_index())
print("Test labels:\n", test["label"].value_counts(normalize=True).sort_index())
```
