# 01 — Data Loading

Load and inspect the cleaned fake-news dataset. This notebook does not train a model.

```python
from pathlib import Path
import pandas as pd

PROJECT_ROOT = Path.cwd()
DATA_PATH = PROJECT_ROOT / "data" / "processed" / "cleaned_news.csv"

if not DATA_PATH.exists():
    raise FileNotFoundError(f"Dataset not found: {DATA_PATH}. See data/README.md.")

df = pd.read_csv(DATA_PATH)
print("Shape:", df.shape)
print("Columns:", df.columns.tolist())
```

```python
required = {"content", "clean_text", "label"}
missing = required - set(df.columns)
if missing:
    raise ValueError(f"Missing required columns: {sorted(missing)}")

print(df.head())
print("\nClass distribution:")
print(df["label"].value_counts().sort_index())
print("\nMissing values:")
print(df[list(required)].isna().sum())
```
